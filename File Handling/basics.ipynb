# Reading a File
Reading a file can be achieved by file.read() which reads the entire content of the file. After reading, it’s good practice to close the file to free up system resources.

```python
file = open("geek.txt", "r") # FileNotFoundError: [Errno 2] No such file or directory: 'geek.txt'
content = file.read() # This will read the entire file and store it in the variable 'content'
print(content)
file.close()
```

# Writing a File
Writing to a file is done using the mode "w". This creates a new file if it doesn’t exist, or overwrites the existing file if it does. The write() method is used to add content. After writing, make sure to close the file.

```python
with open("geek.txt", "w") as file: #Writing to a file (overwrites if file exists)
    file.write("Hello, Python!\n")
    file.write("File handling is easy with Python.")

print("File written successfully")
```

Explanation:

1. "w" mode opens the file for writing (overwrites existing content if the file already exists).
2. write() method adds new text to the file.
3. When using with, the file closes automatically at the end of the block.

# Using with Statement
Instead of manually opening and closing the file, you can use the with statement, which automatically handles closing. This reduces the risk of file corruption and resource leakage.

```python
with open ("geek.txt", "r") as f:
    content = f.read()
    print(content)
print("File read successfully & closed automatically after the block")
```

# Handling Exceptions When Closing a File
It's important to handle exceptions to ensure that files are closed properly, even if an error occurs during file operations. Here, the finally block ensures the file is closed even if an error occurs.

```python
file = None

try:
    file = open("geeqwk.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError as e:
    print("Error:", e)
finally:
    if file is not None:
        file.close() # Finally always executeds, and if say we don't have the file with this name it will give an error when we close the file. So to avoid that problem we check if the file is not None then we close it.  
```
